## Convert all data csv into single csv for dataset preparation - RAW

```python
def _find_filenames(file_dir_path, file_pattern): return list(file_dir_path.glob(file_pattern))
```

```python
file_paths = '/home/salihin/PhD2020/LSTM_blink_eye/BlinkProbDataVal'
datagroup = 'real'
eyeside = 'righteye'
if not isinstance(file_paths, list):
    file_paths = [file_paths]

for file_path in file_paths:
    data_path = Path(file_path)
    filenames = _find_filenames(data_path / f'{datagroup}/', f'*_{eyeside}.csv')
    if eyeside == '*':
        eyeside = ''
    for n, filename in enumerate(filenames):
        print(f'Processing csv files: {n+1}')
        print(filename)
        single_pd = pd.read_csv(filename)
        with open(f'/home/salihin/PhD2020/LSTM_blink_eye/BPDDatasets225/all_{datagroup}_{eyeside}data.csv', 'a') as f:
            write = csv.writer(f)
            write.writerow(single_pd)
```
